```python
file_path = r"C:\Users\immek\GlobalSuperstore.xlsx"

orders = pd.read_excel(file_path, sheet_name="Orders")
people = pd.read_excel(file_path, sheet_name="People")
returns = pd.read_excel(file_path, sheet_name="Returns")
customer = pd.read_excel(file_path, sheet_name="Customer")

print("All datasets loaded successfully!")
```

```python
print("Orders:", orders.shape)
print("People:", people.shape)
print("Returns:", returns.shape)
print("Customer:", customer.shape)
```

```python
datasets = {
    "Orders": orders,
    "People": people,
    "Returns": returns,
    "Customer": customer
}

for name, df in datasets.items():
    print(f"\n{name} Columns:")
    print(df.columns.tolist())
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in datasets.items():
    print(f"\n{name} Dataset:")
    display(df.head())
```

```python
# Check unique IDs in each dataset

print("Orders - Order ID:")
print(orders['Order ID'].nunique())

print("\nOrders - Customer ID:")
print(orders['Customer ID'].nunique())

print("\nCustomer - Customer ID:")
print(customer['Customer ID'].nunique())

print("\nReturns - Order ID:")
print(returns['Order ID'].nunique())
```

```python
for name, df in datasets.items():
    print(f"\n{name} Missing Values:")
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print(f"\n{name} Duplicate Records:")
    print(df.duplicated().sum())
```

```python
# Check Customer ID relationship

print("Customers in Orders:", orders['Customer ID'].nunique())

print("Customers in Customer table:", customer['Customer ID'].nunique())

print("Unmatched Customer IDs:",
      len(set(orders['Customer ID']) - set(customer['Customer ID'])))


# Check Order ID relationship

print("\nOrders in Orders table:", orders['Order ID'].nunique())

print("Orders in Returns table:", returns['Order ID'].nunique())

print("Unmatched Returned Orders:",
      len(set(returns['Order ID']) - set(orders['Order ID'])))
```

```python
# Merge Orders and Customer datasets

orders_customer = pd.merge(
    orders,
    customer,
    on='Customer ID',
    how='left'
)

print("Merged Dataset Shape:", orders_customer.shape)

display(orders_customer.head())
```

```python
# Merge Orders and Returns

orders_returns = pd.merge(
    orders_customer,
    returns,
    on='Order ID',
    how='left'
)

print("Merged Dataset Shape:", orders_returns.shape)

display(orders_returns.head())
```

```python
# Final merged dataset

print("Final Dataset Shape:", orders_returns.shape)

print("\nAll Columns:")
print(orders_returns.columns.tolist())

print("\nMissing Values:")
print(orders_returns.isnull().sum())
```

```python
# Count returned and non-returned orders

return_analysis = orders_returns['Returned'].value_counts(dropna=False)

print("Return Analysis:")
print(return_analysis)
```

```python
return_percentage = (
    orders_returns['Returned'].notna().mean() * 100
)

print("Returned Records Percentage:", round(return_percentage, 2), "%")
```

```python
# Sales analysis based on return status

sales_by_return = orders_returns.groupby(
    orders_returns['Returned'].notna()
)['Sales'].sum()

print(sales_by_return)
```

```python
sales_by_return.index = ['Non-Returned', 'Returned']

print("Sales by Return Status:")
print(sales_by_return)
```

```python
segment_returns = orders_returns.groupby('Segment').agg(
    Total_Orders=('Order ID', 'nunique'),
    Returned_Orders=('Returned', lambda x: x.notna().sum())
)

segment_returns['Return_Rate'] = (
    segment_returns['Returned_Orders'] /
    segment_returns['Total_Orders'] * 100
)

print(segment_returns.sort_values(
    by='Return_Rate',
    ascending=False
))
```

```python
region_returns = orders_returns.groupby('Region').agg(
    Total_Orders=('Order ID', 'nunique'),
    Returned_Orders=('Returned', lambda x: x.notna().sum())
)

region_returns['Return_Rate'] = (
    region_returns['Returned_Orders'] /
    region_returns['Total_Orders'] * 100
)

region_returns = region_returns.sort_values(
    by='Return_Rate',
    ascending=False
)

display(region_returns)
```

```python
category_returns = orders_returns.groupby('Category').agg(
    Total_Orders=('Order ID', 'nunique'),
    Returned_Orders=('Returned', lambda x: x.notna().sum())
)

category_returns['Return_Rate'] = (
    category_returns['Returned_Orders'] /
    category_returns['Total_Orders'] * 100
)

category_returns = category_returns.sort_values(
    by='Return_Rate',
    ascending=False
)

display(category_returns)
```

```python
subcategory_returns = orders_returns.groupby('Sub-Category').agg(
    Total_Orders=('Order ID', 'nunique'),
    Returned_Orders=('Returned', lambda x: x.notna().sum())
)

subcategory_returns['Return_Rate'] = (
    subcategory_returns['Returned_Orders'] /
    subcategory_returns['Total_Orders'] * 100
)

subcategory_returns = subcategory_returns.sort_values(
    by='Return_Rate',
    ascending=False
)

display(subcategory_returns)
```

```python
customer_returns = orders_returns.groupby(
    ['Customer ID', 'Customer Name']
).agg(
    Total_Orders=('Order ID', 'nunique'),
    Returned_Orders=('Returned', lambda x: x.notna().sum())
)

customer_returns['Return_Rate'] = (
    customer_returns['Returned_Orders'] /
    customer_returns['Total_Orders'] * 100
)

customer_returns = customer_returns.sort_values(
    by='Returned_Orders',
    ascending=False
)

display(customer_returns.head(10))
```

```python
segment_sales = orders_returns.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Orders=('Order ID', 'nunique'),
    Total_Profit=('Profit', 'sum')
)

segment_sales = segment_sales.sort_values(
    by='Total_Sales',
    ascending=False
)

display(segment_sales)
```

```python
segment_sales['Profit_Margin'] = (
    segment_sales['Total_Profit'] /
    segment_sales['Total_Sales'] * 100
)

display(segment_sales)
```

```python
region_sales = orders_returns.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique')
)

region_sales = region_sales.sort_values(
    by='Total_Sales',
    ascending=False
)

display(region_sales)
```

```python
region_sales['Profit_Margin'] = (
    region_sales['Total_Profit'] /
    region_sales['Total_Sales'] * 100
)

display(region_sales)
```

```python
country_sales = orders_returns.groupby('Country').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Orders=('Order ID', 'nunique'),
    Total_Customers=('Customer ID', 'nunique')
)

country_sales = country_sales.sort_values(
    by='Total_Sales',
    ascending=False
)

display(country_sales.head(10))
```

```python
country_sales['Profit_Margin'] = (
    country_sales['Total_Profit'] /
    country_sales['Total_Sales'] * 100
)

display(country_sales.head(10))
```

```python
orders_returns['Order Date'] = pd.to_datetime(
    orders_returns['Order Date']
)

orders_returns['Month'] = orders_returns['Order Date'].dt.month_name()

orders_returns['Year'] = orders_returns['Order Date'].dt.year

print(orders_returns[['Order Date', 'Month', 'Year']].head())
```

```python
monthly_sales = orders_returns.groupby(
    ['Year', 'Month']
).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

display(monthly_sales)
```

```python
monthly_sales = orders_returns.groupby(
    ['Year', 'Month']
).agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

display(monthly_sales)
```

```python
monthly_profit = orders_returns.groupby(
    ['Year', 'Month']
).agg(
    Total_Profit=('Profit', 'sum'),
    Total_Sales=('Sales', 'sum')
)

display(monthly_profit)
```

```python
best_month = monthly_profit['Total_Profit'].idxmax()

print("Most Profitable Month:", best_month)

print("Highest Monthly Profit:",
      round(monthly_profit['Total_Profit'].max(), 2))
```

```python
import matplotlib.pyplot as plt

monthly_profit['Total_Profit'].plot(
    kind='bar',
    figsize=(12, 5)
)

plt.title('Monthly Profit Analysis')
plt.xlabel('Year and Month')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
shipping_analysis = orders_returns.groupby('Ship Mode').agg(
    Total_Orders=('Order ID', 'nunique'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

shipping_analysis = shipping_analysis.sort_values(
    by='Total_Sales',
    ascending=False
)

display(shipping_analysis)
```

```python
shipping_analysis['Profit_Margin'] = (
    shipping_analysis['Total_Profit'] /
    shipping_analysis['Total_Sales'] * 100
)

display(shipping_analysis)
```

```python
import matplotlib.pyplot as plt

shipping_analysis['Total_Sales'].plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Sales by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
orders_returns['Ship Date'] = pd.to_datetime(
    orders_returns['Ship Date']
)

orders_returns['Order Date'] = pd.to_datetime(
    orders_returns['Order Date']
)

orders_returns['Shipping_Days'] = (
    orders_returns['Ship Date'] -
    orders_returns['Order Date']
).dt.days

display(
    orders_returns[
        ['Order Date', 'Ship Date', 'Ship Mode', 'Shipping_Days']
    ].head(10)
)
```

```python
shipping_time = orders_returns.groupby('Ship Mode').agg(
    Average_Shipping_Days=('Shipping_Days', 'mean'),
    Minimum_Days=('Shipping_Days', 'min'),
    Maximum_Days=('Shipping_Days', 'max')
)

display(shipping_time.round(2))
```

```python
import matplotlib.pyplot as plt

shipping_time['Average_Shipping_Days'].plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Average Shipping Time by Ship Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Average Shipping Days')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
total_sales = orders_returns['Sales'].sum()

total_profit = orders_returns['Profit'].sum()

total_orders = orders_returns['Order ID'].nunique()

total_customers = orders_returns['Customer ID'].nunique()

total_returns = orders_returns['Returned'].notna().sum()

return_rate = (total_returns / total_orders) * 100

profit_margin = (total_profit / total_sales) * 100

print("GLOBAL SUPERSTORE - BUSINESS REPORT")
print("-----------------------------------")

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Returned Orders:", total_returns)
print("Return Rate:", round(return_rate, 2), "%")
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
top_categories = orders_returns.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

top_categories = top_categories.sort_values(
    by='Total_Sales',
    ascending=False
)

display(top_categories)
```

```python
top_countries = orders_returns.groupby('Country')['Sales'].sum()

top_countries = top_countries.sort_values(
    ascending=False
).head(5)

display(top_countries)
```

```python
print("FINAL BUSINESS INSIGHTS")
print("-----------------------")

print("1. Highest Sales Category:",
      top_categories['Total_Sales'].idxmax())

print("2. Most Profitable Category:",
      top_categories['Total_Profit'].idxmax())

print("3. Highest Sales Country:",
      top_countries.idxmax())

print("4. Most Profitable Month:",
      monthly_profit['Total_Profit'].idxmax())

print("5. Highest Return Rate Region:",
      region_returns['Return_Rate'].idxmax())
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Sales by Category
top_categories['Total_Sales'].plot(
    kind='bar', ax=axes[0, 0]
)
axes[0, 0].set_title('Sales by Category')
axes[0, 0].set_ylabel('Sales')

# 2. Profit by Category
top_categories['Total_Profit'].plot(
    kind='bar', ax=axes[0, 1]
)
axes[0, 1].set_title('Profit by Category')
axes[0, 1].set_ylabel('Profit')

# 3. Sales by Shipping Mode
shipping_analysis['Total_Sales'].plot(
    kind='bar', ax=axes[1, 0]
)
axes[1, 0].set_title('Sales by Shipping Mode')
axes[1, 0].set_ylabel('Sales')

# 4. Return Rate by Region
region_returns['Return_Rate'].plot(
    kind='bar', ax=axes[1, 1]
)
axes[1, 1].set_title('Return Rate by Region')
axes[1, 1].set_ylabel('Return Rate (%)')

plt.tight_layout()
plt.show()
```
